```python
import numpy as np

svalues_nucl = []
svalues_cyto = []

for i in range(1, 11):
    with open(f"batches/s_batch{i:02d}.txt") as f:
        row = f.readline().split()
        svalues_cyto.append(float(row[3]))   # S_cito
        svalues_nucl.append(float(row[5]))   # S_nucleo

svalues_cyto = np.array(svalues_cyto)
svalues_nucl = np.array(svalues_nucl)

mean_cyto = svalues_cyto.mean()
sem_cyto  = svalues_cyto.std(ddof=1) / np.sqrt(len(svalues_cyto))

mean_nucl = svalues_nucl.mean()
sem_nucl  = svalues_nucl.std(ddof=1) / np.sqrt(len(svalues_nucl))

print(f"S(C<-N) = {mean_cyto:.4f} ± {sem_cyto:.4f} mGy/Bq.s")
print(f"S(N<-N) = {mean_nucl:.4f} ± {sem_nucl:.4f} mGy/Bq.s")
```
